```python
import os
import torch
import torchaudio
from transformers import Wav2Vec2ForCTC, Wav2Vec2Processor
from google.colab import files

print("Please upload your .wav files")
uploaded = files.upload()

wav_directory = "/content/wavs"
if not os.path.exists(wav_directory):
    os.makedirs(wav_directory)


for filename in uploaded.keys():
    os.rename(filename, os.path.join(wav_directory, filename))

out_file = os.path.join(wav_directory, "metadata.txt")


file_and_transcripts = []


model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-large-960h")
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-large-960h")


for filename in os.listdir(wav_directory):
    if filename.endswith(".wav"):
        wav_file = os.path.join(wav_directory, filename)

        try:
            waveform, sample_rate = torchaudio.load(wav_file)
            waveform = waveform.mean(dim=0) if waveform.ndimension() > 1 else waveform
            if sample_rate != 16000:
                resampler = torchaudio.transforms.Resample(orig_freq=sample_rate, new_freq=16000)
                waveform = resampler(waveform)

            input_values = processor(waveform, return_tensors="pt", sampling_rate=16000).input_values

            with torch.no_grad():
                logits = model(input_values).logits
            predicted_ids = torch.argmax(logits, dim=-1)
            transcript = processor.decode(predicted_ids[0])
            file_and_transcripts.append(f"/content/wavs/{filename}|{transcript}")

        except Exception as e:
            print(f"Error processing {wav_file}: {e}")
            continue

with open(out_file, "w") as f:
    for line in file_and_transcripts:
        f.write(f"{line}\n")

print(f"Output file '{out_file}' created successfully.")

files.download(out_file)
```
