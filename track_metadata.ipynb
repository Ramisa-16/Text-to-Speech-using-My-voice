```python
import os
import shutil
import taglib
from google.colab import files

print("Please upload your .wav files")
uploaded = files.upload()
input_folder = "/content"
output_folder = "/content/titled_wavs"

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

def update_metadata(input_folder, output_folder, n):
    for i in range(1, n+1):
        input_file = os.path.join(input_folder, f'{i}.wav')
        output_file = os.path.join(output_folder, f'{i}.wav')

        if os.path.exists(input_file):
            with taglib.File(input_file) as audio:
                audio.tags["TITLE"] = [f'{i}']
                audio.tags["TRACKNUMBER"] = [f'{i}']
                audio.save()

            shutil.copy2(input_file, output_file)
            print(f"Updated metadata for {i}.wav: title='{i}', track number={i}")

        else:
            print(f"File {i}.wav not found.")

n = len(uploaded)
update_metadata(input_folder, output_folder, n)

import shutil
shutil.make_archive('/content/titled_wavs', 'zip', output_folder)
files.download('/content/titled_wavs.zip')
```
