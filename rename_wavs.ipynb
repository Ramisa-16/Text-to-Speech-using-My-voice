```python
import os
from google.colab import files

def rename_wavs(folder_path):
    files = os.listdir(folder_path)
    wavs = [f for f in files if f.lower().endswith('.wav')]

    print(f'Folder path: {folder_path}')
    print(f'All files in the folder: {files}')
    print(f'Total Wave files found: {len(wavs)}')

    for index, wav in enumerate(wavs, start=1):
        old_path = os.path.join(folder_path, wav)
        new_path = os.path.join(folder_path, f'{index}.wav')
        os.rename(old_path, new_path)
        print(f'Renamed {old_path} to {new_path}')


if __name__ == "__main__":
    print("Upload your .wav files")
    uploaded = files.upload()

    folder_path = "/content/wavs"
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    for file_name in uploaded.keys():
        os.rename(file_name, os.path.join(folder_path, file_name))
    rename_wavs(folder_path)
    print("Files renamed!!!!!!!")
```
